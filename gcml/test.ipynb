```python
import tensorflow as tf
import sklearn

import time

print("GPU disponibili:", tf.config.list_physical_devices('GPU'))
print("Versione TensorFlow:", tf.__version__)
print("Versione scikit-learn:", sklearn.__version__)
```

```python


# monitora da terminale con: nvidia-smi -l 1
x = tf.random.normal([10000, 10000])
y = tf.random.normal([10000, 10000])

print("Inizio calcolo...")
start = time.time()
z = tf.matmul(x, y)
end = time.time()

print(f"Tempo calcolo: {end - start:.3f}s")
print(f"Risultato: {z[0, 0]}")
```
